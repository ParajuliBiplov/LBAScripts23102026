### Excel generation to list out all files within a Folder

```python
import os
import pandas as pd

def list_filenames_to_excel(directory, output_excel_file):
    # Get list of files in directory
    files = os.listdir(directory)
    
    # Filter out directories
    files = [f for f in files if os.path.isfile(os.path.join(directory, f))]
    
    # Create a DataFrame
    df = pd.DataFrame(files, columns=['Filename'])
    
    # Write DataFrame to Excel file
    df.to_excel(output_excel_file, index=False)
    
    print(f"Excel file '{output_excel_file}' created with {len(files)} filenames.")

# Define the directory and output file path
directory = r'L:\LBDB_ZWS_Finalized\WithSortie_Finalized\106G_2842'
output_excel_file = r'L:\LBDB_ZWS_Finalized\WithSortie_Finalized\filenames.xlsx'


# Call the function
list_filenames_to_excel(directory, output_excel_file)
```

### Excel generation to list out all Folders 

```python
import os
import pandas as pd

def list_folders_to_excel(directory, output_excel_file):
    # Get list of items in directory
    items = os.listdir(directory)
    
    # Filter only directories (folders)
    folders = [f for f in items if os.path.isdir(os.path.join(directory, f))]
    
    # Create a DataFrame
    df = pd.DataFrame(folders, columns=['FolderName'])
    
    # Write DataFrame to Excel file
    df.to_excel(output_excel_file, index=False)
    
    print(f"Excel file '{output_excel_file}' created with {len(folders)} folders.")

# Define the directory and output file path
directory = r'L:\LBDB_ZWS_Finalized\WithSortie_Finalized'
output_excel_file = r'L:\LBDB_ZWS_Finalized\WithSortie_Finalized\foldernames.xlsx'

# Call the function
list_folders_to_excel(directory, output_excel_file)
```

### Excel generation to list out all images in a Path 

```python
from pathlib import Path
import pandas as pd
from tqdm import tqdm

def list_images_to_excel(directory, output_excel_file, recursive=True):
    try:
        directory = Path(directory)

        image_exts = {".jpg", ".jpeg", ".png", ".bmp", ".tiff", ".tif", ".gif", ".webp"}

        # collect all files first
        if recursive:
            all_files = list(directory.rglob("*"))
        else:
            all_files = list(directory.iterdir())

        image_files = []

        # progress bar while filtering
        for f in tqdm(all_files, desc="Scanning files", unit="file"):
            if f.is_file() and f.suffix.lower() in image_exts:
                image_files.append(f)

        # build dataframe
        df = pd.DataFrame({
            "Filename": [f.name for f in image_files],
            "FullPath": [str(f) for f in image_files],
            "Extension": [f.suffix.lower() for f in image_files],
        })

        # saving progress (simple indicator)
        print("Writing to Excel...")

        df.to_excel(output_excel_file, index=False)

        print(f"Done. {len(image_files)} images saved.")

    except PermissionError:
        print("Permission denied (check W: or output folder access).")
    except FileNotFoundError:
        print("Directory not found.")
    except Exception as e:
        print(f"Error: {e}")

directory = r'W:'
output_excel_file = r'G:\0 BIPLOV\TASKS\Patrick\ID0000Problem\ListofAllImages.xlsx'

list_images_to_excel(directory, output_excel_file, recursive=True)
```
