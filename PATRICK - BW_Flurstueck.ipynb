```python
import pandas as pd
import requests
import os
```

```python
# Specify the path to the CSV file
file_path = r'D:\BIPLOV\Reference GeoData Preparation_Bundesland\BW\Flurstueck\Processing\Gemarkungen.csv'

# Read the CSV file into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows of the DataFrame to verify it's read correctly
df.head()
```

```python
# Convert gemarkung_id to string and pad with zeros to ensure 4-digit length
df['gemarkung_id'] = df['gemarkung_id'].astype(str).str.zfill(4)

# Sort DataFrame by gemarkung_id in ascending order
df.sort_values(by='gemarkung_id', inplace=True)

# Export the modified DataFrame to 'Gemarkungen_modified.csv' without index
output_file = r'D:\BIPLOV\Reference GeoData Preparation_Bundesland\BW\Flurstueck\Processing\Gemarkungen_modified.csv'
df.to_csv(output_file, index=False, encoding='utf-8-sig')

# Display the modified DataFrame
df.head(100)
```

```python
# Generate the full URL for downloading
# df['download_url'] = 'https://opengeodata.lgl-bw.de/data/alkis/' + df['filename'].str.replace(' ', '_').str.replace('/', '_')
df['download_url'] = 'https://opengeodata.lgl-bw.de/data/alkis/' + df['filename']


# Function to download file
def download_file(url, save_path):
    with requests.get(url, stream=True) as r:
        r.raise_for_status()
        with open(save_path, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)

# Directory to save downloaded files
download_dir = r'D:\BIPLOV\Reference GeoData Preparation_Bundesland\BW\Flurstueck\Processing\Downloads'

# Create the directory if it doesn't exist
os.makedirs(download_dir, exist_ok=True)

# Download each file
for index, row in df.iterrows():
    url = row['download_url']
    filename = row['filename']
    save_path = os.path.join(download_dir, filename)
    print(f"Downloading {filename}...")
    download_file(url, save_path)
    print(f"{filename} downloaded successfully.")

print("All files downloaded successfully.")
```
