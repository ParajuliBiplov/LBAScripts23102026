```python
import geopandas as gpd
import os
import glob
import pandas as pd
from datetime import datetime
```

# Saarland

```python
# Step 1: Define the input directory and output file
input_dir = r"T:\0 BIPLOV\Reference GeoData Preparation_Bundesland\ALKIS"
output_file = os.path.join(input_dir, 'Flurstueck_Saarland.gpkg')

# Step 2: Find all GeoPackage files in the input directory
gpkg_files = glob.glob(os.path.join(input_dir, "*.gpkg"))

# Step 3: Function to merge GeoPackage files
def merge_geopackages(gpkg_files, output_file):
    # Initialize the list for storing GeoDataFrames
    gdfs = []
    for gpkg_file in gpkg_files:
        try:
            gdf = gpd.read_file(gpkg_file)
            gdfs.append(gdf)
        except Exception as e:
            print(f"Failed to read {gpkg_file}: {e}")
    
    # Concatenate all GeoDataFrames
    if gdfs:  # Check if the list is not empty
        merged_gdf = gpd.GeoDataFrame(pd.concat(gdfs, ignore_index=True))
        
        # Save the merged GeoDataFrame to a single GeoPackage file
        merged_gdf.to_file(output_file, layer='cadastral_parcels', driver="GPKG")
        print(f"Merged GeoPackage saved to {output_file}")
    else:
        print("No GeoDataFrames to merge.")

# Step 4: Record the initial time
initial_time = datetime.now()
print(f"Initial time: {initial_time}")

# Step 5: Merge the GeoPackage files
merge_geopackages(gpkg_files, output_file)

# Step 6: Record the end time
end_time = datetime.now()
print(f"End time: {end_time}")

# Step 7: Calculate and print the duration
duration = end_time - initial_time
print(f"Duration: {duration}")
```

# BW

```python
# Step 1: Define the input directory and output file
input_dir = r"D:\BIPLOV\Reference GeoData Preparation_Bundesland\BW\Flurstueck\Processing\Downloads\Extracted"
output_file = os.path.join(input_dir, 'Flurstueck_Combined_BW_CRS32N.gpkg')

# Step 2: Find all shapefile (.shp) files in the input directory
shp_files = glob.glob(os.path.join(input_dir, "**", "flurstueck.shp"), recursive=True)

# Set the target CRS (assuming ETRS89 / UTM zone 32N)
target_crs = "EPSG:25832"

# Step 3: Function to merge shapefiles and save as a GeoPackage
def merge_shapefiles_to_geopackage(shp_files, output_file):
    gdfs = []
    
    for idx, shp_file in enumerate(shp_files, start=1):
        try:
            gdf = gpd.read_file(shp_file)
            
            # Check if the CRS is missing, and set it to the target CRS
            if gdf.crs is None:
                print(f"CRS missing for {shp_file}. Setting CRS to {target_crs}.")
                gdf.set_crs(target_crs, inplace=True)
            
            # Reproject to the target CRS if necessary
            elif gdf.crs != target_crs:
                print(f"Reprojecting {shp_file} to {target_crs}.")
                gdf = gdf.to_crs(target_crs)
            
            gdfs.append(gdf)
            print(f"Shapefile {idx}/{len(shp_files)} read: {shp_file}")
        
        except Exception as e:
            print(f"Failed to read {shp_file}: {e}")
    
    print(f"Total number of shapefiles read: {len(gdfs)}")
    
    if gdfs:
        merged_gdf = gpd.GeoDataFrame(pd.concat(gdfs, ignore_index=True), crs=target_crs)
        merged_gdf.to_file(output_file, layer='cadastral_parcels', driver="GPKG")
        print(f"Merged GeoPackage saved to {output_file}")
    else:
        print("No shapefiles to merge.")

# Step 4: Record the initial time
initial_time = datetime.now()
print(f"Initial time: {initial_time}")

# Step 5: Merge the shapefiles into a GeoPackage
merge_shapefiles_to_geopackage(shp_files, output_file)

# Step 6: Record the end time
end_time = datetime.now()
print(f"End time: {end_time}")

# Step 7: Calculate and print the duration
duration = end_time - initial_time
print(f"Duration: {duration}")
```
