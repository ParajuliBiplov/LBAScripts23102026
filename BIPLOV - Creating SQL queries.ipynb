### Creating SQL queries for all SORTIEs with same Date in W:\

```python
import os

# Input from user
date_input = input("Enter date (YYYY-MM-DD or YYYY_MM_DD format): ").strip().replace("-", "_")

# Path and date setup
PATH_A = r"W:\\"
target_date = date_input

# Collect SORTIE values
sortie_values = []

try:
    with os.scandir(PATH_A) as entries:
        for entry in entries:
            if entry.is_dir():
                folder_name = entry.name
                
                # Check if folder starts with the date
                if folder_name.startswith(target_date):
                    # Split using '#' as delimiter
                    parts = folder_name.split('#')
                    
                    # We need at least 3 parts: date, empty, sortie
                    if len(parts) >= 3:
                        sortie_part = parts[2].strip()  # Third part is the SORTIE
                        
                        # Check if sortie_part is not just underscores or empty
                        if sortie_part and not all(c == '_' or c == ' ' for c in sortie_part):
                            # Remove leading underscores if present
                            if sortie_part.startswith('_'):
                                sortie_part = sortie_part[1:]
                            
                            if sortie_part:  # Check again after cleaning
                                sortie_values.append(sortie_part)

except (FileNotFoundError, PermissionError) as e:
    print(f"Error accessing path: {e}")

# Remove duplicates and sort
sortie_values = sorted(set(sortie_values))

# Generate only the IN clause
if sortie_values:
    # Create IN clause exactly in the specified format
    sortie_in_clause = "', '".join(sortie_values)
    
    # Output only the IN clause part
    print(f"sortie_ori IN ('{sortie_in_clause}')")
else:
    print(f"No SORTIE values found for date: {target_date}")
```
