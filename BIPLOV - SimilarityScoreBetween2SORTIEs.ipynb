### Similarity matching 

```python
import pandas as pd
from rapidfuzz import process, fuzz
from datetime import datetime
import os

# File path
file_path = r"G:\0 BIPLOV\TASKS\BIPLOV\SortieFIX_Bilderbestellungen_fluginfos_ImportBilder\SortieFIX_Bilderbestellungen_fluginfos_ImportBilder.xlsx"

# Load sheets
df_import = pd.read_excel(file_path, sheet_name='BF_ImportBilder')
df_db = pd.read_excel(file_path, sheet_name='BF_DB')

# Extract SORTIE columns (column A)
sorties_import = df_import.iloc[:, 0].dropna().astype(str)
sorties_db = df_db.iloc[:, 0].dropna().astype(str)

# Prepare results list
results = []

# Matching loop
for sortie in sorties_import:
    match, score, _ = process.extractOne(
        sortie,
        sorties_db,
        scorer=fuzz.ratio  # you can try fuzz.partial_ratio if needed
    )
    
    results.append({
        "SORTIE_Import": sortie,
        "Best_Match_DB": match,
        "Similarity_%": score
    })

# Convert to DataFrame
df_results = pd.DataFrame(results)

# Create timestamp
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

# Output path
output_file = os.path.join(
    r"G:\0 BIPLOV\TASKS\BIPLOV",
    f"Sortie_Matching_Report_{timestamp}.xlsx"
)

# Save to Excel
df_results.to_excel(output_file, index=False)

print(f"Report saved to: {output_file}")
```
