### DEM Mosaicing

```python
import requests
import geopandas as gpd
from shapely.geometry import shape


import rasterio
from rasterio.io import MemoryFile
from shapely.geometry import box
import geopandas as gpd
import os
from rasterio.transform import from_bounds
import glob
import pandas as pd
from datetime import datetime
import time
from osgeo import gdal
```

```python
# Define the input directory containing the TIF files
input_dir = r"D:\BIPLOV\Reference GeoData Preparation_Bundesland\HESSEN\DTM\DGM\DGM_HESSEN"
output_file = os.path.join(input_dir, 'DEM_mosaic_HESSEN.tif')

# Record the initial time
initial_time = datetime.now()
print(f"Initial time: {initial_time}")

# Find all TIF files in the input directory
tif_files = glob.glob(os.path.join(input_dir, "*.tif"))

# Create a VRT (Virtual Dataset) from the TIF files with nodata value set to 0
vrt_options = gdal.BuildVRTOptions(resampleAlg='nearest', srcNodata=0, VRTNodata=0)
vrt = gdal.BuildVRT('/vsimem/temporary.vrt', tif_files, options=vrt_options)

# Translate the VRT to a TIF file (mosaic)
gdal.Translate(output_file, vrt)

# Clean up the VRT
gdal.Unlink('/vsimem/temporary.vrt')

# Record the end time
end_time = datetime.now()
print(f"End time: {end_time}")

# Calculate and print the duration
duration = end_time - initial_time
print(f"Duration: {duration}")

print(f"Mosaic created and saved to {output_file}")
```
